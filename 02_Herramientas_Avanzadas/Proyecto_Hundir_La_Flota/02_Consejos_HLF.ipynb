1. Crea un tablero de 10x10 relleno del carácter espacio " ". (Utiliza un array de Numpy para representarlo y creálo utilizando las inicializaciones que hemos aprendidio)

```python
def crear_tablero(lado = 10, agua = " "):
    tablero = np.full((lado, lado), agua)
    return tablero
```

```python
tablero = crear_tablero()
print(tablero)
```

2. Posiciona un par de barcos en [(0,1), (1,1)] y [(1,3), (1,4), (1,5), (1,6)]. Los barcos serán Os mayúsculas. Como ves, un barco de dos posiciones de eslora y otro de cuatro.

```python
barco1 = [(0,1),(1,1)]

def colocar_barco(tablero, barco):
    for pieza in barco:
        tablero[pieza] = "O"
    return tablero

tablero = colocar_barco(tablero, barco1)
print(tablero)
```

3. Recibe un disparo en uno de los barcos, sustituyendo la O por una X
4. Recibe un disparo en agua, sustituyendo uno de los espacios por un guión: -

```python
def realizar_disparo(tablero, coordenada):
    if tablero[coordenada] == "O":
        tablero[coordenada] = "X"
        print("¡Tocado!")
    elif tablero[coordenada] == " ":
        tablero[coordenada] = "-"
        print("Agua.")
    elif tablero[coordenada] == "X" or tablero[coordenada] == "-":
        print("Ya has disparado aquí.")
```

```python
realizar_disparo(tablero, (0,1))
print(tablero)
```

```python
# Opción Swith (match/case) en Python

def realizar_disparo_switch(disparo, tablero):
    match tablero[disparo]:
        case "O":
            tablero[disparo] = "X"
            print("¡Tocado!")
        case " ":
            tablero[disparo] = "-"
            print("Agua.")
        case "X", "-":
            print("Ya has disparado aquí.")
```

5. Ignorando los disparos recibidos, ahora prueba a posicionar otro barco de 4 posiciones dentro del tablero, pero **de manera ALEATORIA.**

**TIP apartado 5**: El barco tendrá una posición inicial aleatoria dentro del tablero, y una orientación también aleatoria (N,S,E,O).

Por ejemplo, si tu posicion inicial fuese (5,0) y la orientación O, no podrias ponerlo porque se sale del tablero.  

Ahora bien, si fuese orientación Este, si podrías, y ocuparía las posiciones [(5,0), (5,1), (5,2), (5,3)]. Por tanto tendrás que iterar con un bucle si es posible encajar el barco en el tablero, y el bucle terminará cuando lo consigas encajar. **Ojo con las posiciones donde ya hay barco!!!**

Usa `import random` para elegir las coordenadas aleatorias.

```python
# Un buen enfoque consiste en dividir el problema en problemas más pequeños.
# Primero definiremos una función que pueda colocar barcos comprobando que no se salgan del tablero o coincidan con barcos ya existentes.
# Después definiremos la función que pueda posicionar de forma aleatoria.
# Aprovechando que las funciones pueden llamar a otras funciones: nuestra función de posición aleatoria hará uso de la función de colocar barcos.
```

```python
def colocar_barco_plus(tablero, barco):
    tablero_temp = tablero.copy()
    indice_max_filas = tablero.shape[0] - 1
    indice_max_columnas = tablero.shape[1] - 1

    for pieza in barco:
        fila = pieza[0]
        columna = pieza[1]

        if fila < 0  or fila > indice_max_filas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if columna < 0 or columna > indice_max_columnas:
            print(f"No puedo poner la pieza {pieza} porque se sale del tablero")
            return False
        if tablero[pieza] == "O" or tablero[pieza] == "X":
            print(f"No puedo poner la pieza {pieza} porque hay otro barco")
            return False
        
        tablero_temp[pieza] = "O"

    return tablero_temp
```

```python
def crea_barco_aleatorio(tablero, eslora = 4, num_intentos = 100):
    indice_max_filas = tablero.shape[0] - 1
    indice_max_columnas = tablero.shape[1] - 1
    contador = 0

    while contador <= num_intentos:        
        contador += 1

        barco = []
        pieza_original = (random.randint(0, indice_max_filas), random.randint(0, indice_max_columnas))
        print("Pieza original:", pieza_original)
        barco.append(pieza_original)

        orientacion = random.choice(["N","S","O","E"])
        print("Con orientacion", orientacion)

        fila = pieza_original[0]
        columna = pieza_original[1]

        for i in range(eslora - 1):
            if orientacion == "N":
                fila -= 1
            elif orientacion  == "S":
                fila += 1
            elif orientacion == "O":
                columna -= 1
            elif orientacion == "E":
                columna += 1
            pieza = (fila,columna)
            barco.append(pieza)
            
        tablero_temp = colocar_barco_plus(tablero, barco)

        if type(tablero_temp) == np.ndarray:
            print("Barco colocado, número de intentos:", contador)
            return tablero_temp
        
        print("Tengo que intentar colocar otro barco")

```

```python
tablero = crea_barco_aleatorio(tablero)
print(tablero)
```
