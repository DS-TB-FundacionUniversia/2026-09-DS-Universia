# Web scraping con Beautiful Soup

El **web scraping** es el proceso de extracción de datos de sitios web de manera automatizada. Aprenderás los conceptos fundamentales y las herramientas necesarias para recolectar datos valiosos de la web, lo que te permitirá alimentar tus análisis y proyectos de Data Science con información fresca y relevante.

## ¿Qué es?

En términos sencillos, web scraping es como recolectar datos de una página web. Imagina que la web es una mina de información y que eres un minero recogiendo las gemas que necesitas para tu proyecto. En lugar de copiar y pegar manualmente, utilizamos código para automatizar el proceso.

### ¿Por qué es importante el Web Scraping?


* **Acceso a datos no estructurados**: La mayoría de los datos en línea no están disponibles en formatos estructurados y listos para ser analizados. El web scraping te permite convertir datos no estructurados en un formato que puedas utilizar.

* **Actualización de datos**: Muchas fuentes web se actualizan constantemente. El web scraping te permite mantener tus conjuntos de datos actualizados sin tener que hacerlo manualmente.

* **Amplia gama de aplicaciones**: Desde la recopilación de datos de redes sociales hasta la obtención de precios de productos, el web scraping se utiliza en una amplia gama de aplicaciones de Data Science.

## Proceso de Web Scraping

El web scraping generalmente sigue estos pasos:

* **Hacer una solicitud HTTP**: obtenemos el contenido HTML de una página web.

* **Análisis del HTML**: analizamos la estructura del HTML y encontramos los elementos que contienen los datos que necesitamos.

* **Extracción de datos**: una vez identificados los elementos, extraemos los datos relevantes.

* **Almacenamiento de datos**: guardamos los datos en un formato estructurado, como un archivo CSV o una base de datos.

* **Limpieza y procesamiento**: a menudo, los datos extraídos necesitan ser limpiados y preprocesados antes de su análisis.

### Aspectos legales y éticos

El web scraping puede tener implicaciones legales y éticas. Antes de scrapear una web, revisa:

* **`robots.txt`**: casi todas las webs publican en `https://dominio/robots.txt` qué rutas permiten (`Allow`) o prohíben (`Disallow`) a los robots. No es una ley, pero es la primera señal de lo que el sitio acepta.  

* **Términos de uso**: algunos sitios prohíben expresamente la extracción automatizada de sus datos.
* **Volumen de peticiones**: no sobrecargues el servidor. Haz pausas entre peticiones (`time.sleep`) y descarga solo lo que necesitas.
* **Datos personales**: si extraes datos de personas (nombres, correos, perfiles...), en la UE se aplica el **RGPD**, aunque esos datos sean públicos.
* **Propiedad intelectual**: que un contenido sea accesible no significa que puedas redistribuirlo.
* **¿Hay una API?**: si el sitio ofrece una API oficial, úsala. Es más estable, más eficiente y está permitida.

## Intro HTML

HTML (HyperText Markup Language) es el lenguaje de marcado utilizado para crear páginas web. Comprender los conceptos básicos de HTML es esencial para realizar web scraping, ya que te permite navegar y extraer datos de manera efectiva de sitios web.

HTML utiliza etiquetas para estructurar el contenido de una página web. Aquí tienes una breve introducción a los conceptos básicos de HTML:

1. **Etiquetas:** Las etiquetas son elementos fundamentales en HTML. Comienzan con `<` y terminan con `>`. Por ejemplo, `<p>` se utiliza para definir un párrafo, `<h1>` para encabezados de nivel 1, y `<a>` para enlaces.

2. **Elementos:** Un elemento HTML se compone de una etiqueta de apertura, contenido y una etiqueta de cierre. Por ejemplo, `<p>Este es un párrafo.</p>`. El contenido es lo que se muestra en la página.

3. **Atributos:** Las etiquetas pueden contener atributos que proporcionan información adicional sobre el elemento. Por ejemplo, en `<a href="https://www.ejemplo.com">Enlace</a>`, "href" es un atributo que define la URL a la que el enlace apunta.

4. **Anidamiento:** Los elementos HTML pueden anidarse dentro de otros elementos. Por ejemplo, un párrafo `<p>` puede contener un enlace `<a>`. Es importante entender la jerarquía de anidamiento para navegar eficazmente por la estructura de una página web.

5. **Estructura:** Una página web típica tiene una estructura jerárquica con elementos como `<html>`, `<head>` (que contiene metadatos), `<body>` (que contiene el contenido visible), y muchos otros elementos como encabezados, listas, imágenes, etc.

6. **Clases e IDs:** Los elementos HTML pueden tener clases y IDs, que son atributos utilizados para aplicar estilos y facilitar la identificación de elementos específicos en una página. Estos son útiles al realizar web scraping para seleccionar elementos específicos.

Para realizar web scraping, necesitas comprender cómo se organiza y etiqueta la información en una página web para poder identificar y extraer los datos que necesitas. El conocimiento de HTML es una base fundamental para esta tarea, ya que te permite navegar y seleccionar elementos específicos en el código fuente de una página web.

### Ejemplo con un HTML sencillo

Antes de ir a una web real, practiquemos con un HTML pequeño escrito por nosotros. Así vemos las herramientas básicas de BeautifulSoup sobre un documento que **no cambia**, mientras que las webs reales modifican su estructura con frecuencia.

```python
from bs4 import BeautifulSoup as bs

html_ejemplo = """
<html>
  <head><title>Mi tienda de libros</title></head>
  <body>
    <h1 id="titulo-principal">Libros destacados</h1>
    <ul id="libros">
      <li class="libro"><a href="/libros/1">Cien años de soledad</a> <span class="precio">12,50</span></li>
      <li class="libro"><a href="/libros/2">Don Quijote de la Mancha</a> <span class="precio">9,95</span></li>
      <li class="libro"><a href="/libros/3">La sombra del viento</a> <span class="precio">15,00</span></li>
    </ul>
  </body>
</html>
"""

soup_ejemplo = bs(html_ejemplo, "html.parser")

print(soup_ejemplo.title.get_text())             # texto de la etiqueta <title>
print(soup_ejemplo.find("h1")["id"])              # atributo id del <h1>
print(soup_ejemplo.find("ul", id="libros").name)  # buscar por etiqueta + atributo

for libro in soup_ejemplo.find_all("li", class_="libro"):
    titulo = libro.find("a").get_text()
    enlace = libro.find("a")["href"]
    precio = float(libro.find("span", class_="precio").get_text().replace(",", "."))
    print(titulo, enlace, precio)
```

### BeautifulSoup

Beautiful Soup es una popular biblioteca de Python utilizada para analizar y extraer información de documentos HTML y XML de una manera sencilla y eficaz. Esta biblioteca proporciona herramientas para navegar y buscar elementos dentro de la estructura de un documento web, lo que la convierte en una herramienta esencial para tareas de web scraping y procesamiento de datos web. Beautiful Soup facilita la extracción de datos de una página web al proporcionar una interfaz amigable que permite acceder a etiquetas, atributos y contenido de manera intuitiva, lo que la convierte en una elección común entre los científicos de datos y desarrolladores que trabajan con datos web.

```python
from bs4 import BeautifulSoup as bs
import requests
import pandas as pd
import numpy as np

pd.set_option('max_colwidth', 800)
```

Ahora estamos listos para solicitar nuestra primera página web. No es nada complicado: guardamos la URL que queremos scrapear en la variable `url`, hacemos la petición con `requests.get(url)` y guardamos la respuesta en la variable `response`.

Conviene acostumbrarse desde el principio a tres buenas prácticas:

* **`headers` con `User-Agent`**: por defecto `requests` se identifica como `python-requests/x.y`, y muchas webs bloquean ese agente (respuesta 403). Enviar un `User-Agent` de navegador evita muchos de esos bloqueos.
* **`timeout`**: sin él, si el servidor no responde, la petición puede quedarse colgada indefinidamente.
* **`response.raise_for_status()`**: si la respuesta es un error (403, 404, 429...), lanza una excepción en lugar de seguir trabajando con una página de error sin darnos cuenta.

Además, si vas a hacer **varias peticiones** (por ejemplo, recorrer varias páginas), añade una pausa entre ellas con `time.sleep(1)` o similar para no sobrecargar el servidor ni ser bloqueado.

```python
url = "https://www.filmaffinity.com/es/topgen.php"
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/124.0 Safari/537.36"}

response = requests.get(url, headers=headers, timeout=10)
response.raise_for_status()  # lanza una excepción si el código no es 2xx (403, 404, 429, 500...)
```

```python
print(response)
```

Pero necesitamos el contenido HTML de la página web solicitada, así que como siguiente paso guardamos el contenido de la respuesta en la variable `html`:

### **Cuidado: emparejar listas por posición**

Las dos soluciones anteriores obtienen los títulos y las notas con **dos `find_all` independientes** y las juntan por posición. Parece que funciona, pero **el resultado es incorrecto**: fíjate en que hay 30 películas y, sin embargo, la celda que imprime todos los `div.avg` muestra 90 notas. FilmAffinity repite cada nota 3 veces en el HTML (versión escritorio, versión móvil y ficha), así que la lista de notas queda `[9.0, 9.0, 9.0, 8.9, 8.9, 8.9, ...]` y a *Planeta Tierra II* le asignamos el 9,0 de *El padrino*. Y lo peor: **no salta ningún error**.

La forma robusta es localizar primero el **contenedor de cada película** (`div.movie-card`) y extraer todos sus campos desde ese mismo nodo. Así título y nota siempre pertenecen a la misma película, y si falta un campo lo dejamos como `None` en lugar de desplazar el resto.

```python
top10 = {"Titulo": [],
         "Rating": []}

for card in soup.find_all("div", class_="movie-card")[:10]:
    titulo = card.find("div", class_="mc-title").find("a").get_text(strip=True)
    nota = card.find("div", class_="avg")

    top10["Titulo"].append(titulo)
    top10["Rating"].append(float(nota.get_text().replace(",", ".")) if nota else None)

df10 = pd.DataFrame(top10, index=range(1, 11))
df10.to_csv("data/top10.csv")
df10
```
