### Ejercicio 9

Prueba ahora a hacer una feature selection utilizando ANOVA. Compara la validación cruzada con la del modelo baseline y con la del modelo anterior. (Pista: SelectKBest)

*La cantidad de features ideal con ANOVA normalmente se obtiene luego de probar con varias cantidades, si tenemos una intuición de cuál puede ser el número ideal tras el análisis de datos, también podemos tirar de ella; el objetivo siempre será reducir la máxima cantidad de features con el menor coste en cuanto a rendimiento. En este caso podemos usar 5 por simplicidad. Aplicamos ANOVA con el f-test ya que se adapta a nuestra problemática de utilizar variables continuas para predecir un target categórico. (Por simplicidad, seleccionamos con todo el train y luego validamos; como vimos en la teoría, lo riguroso sería meter el selector en un `Pipeline` para que se reajuste en cada fold.)*

```python
# Encontrar las 5 mejores características
num_feats_anova = 5
select_feature = SelectKBest(f_classif, k=num_feats_anova).fit(X_train, y_train)
```

*Fíjate en el uso del atributo `scores_` de SelectKBest para ordenar las variables:*

```python
best_features_sorted = pd.DataFrame({'Feature': X_train.columns, 'Score': select_feature.scores_}).sort_values(by='Score', ascending=False)
features_anova = select_feature.get_feature_names_out()
best_features_sorted.head()
```

```python
print(features_anova)
```

*Las mejores features según ANOVA son las que acabamos de imprimir: medidas del perímetro, del radio y de los puntos cóncavos (varias de ellas en su versión "worst"). Evaluamos el modelo con ellas.*

```python
# Transformar el dataset a solo las 5 mejores características
X_train_anova = select_feature.transform(X_train)


#Random Forest Classifier
rfc_2 = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=5)

anova_score = np.mean(cross_val_score(rfc_2, X_train_anova, y_train, cv = 5, scoring= "balanced_accuracy"))

print(f"ANOVA: {anova_score*100.0}%/{num_feats_anova} feats. vs Baseline: {baseline*100.0}%/{num_feats_baseline} feats.")
```

*Perdemos un poco (92,4% frente a 93,9%), pero con solo 5 de 30 features. No está mal.*

*Comparemos features con nuestro análisis visual*

```python
print([col for col in features_anova if col not in features_visual_final])
```

### Ejercicio 10

Prueba a hacer ahora una selección de features con RFE (Recursive Feature Elimination), hasta quedarte con 5 features. Evalúa el rendimiento en "balanced_accuracy" con validación cruzada y compara con el baseline.

```python
from sklearn.feature_selection import RFE

#Random Forest Classifier
rfc_3 = RandomForestClassifier(n_estimators=100, random_state=42, max_depth=5)

#Recursive Feature Elimination y entrenar el modelo
num_feats_rfe = 5
rfe = RFE(estimator=rfc_3, n_features_to_select= num_feats_rfe, step=1)
rfe = rfe.fit(X_train, y_train)
X_train_rfe = rfe.transform(X_train)

rfe_score = np.mean(cross_val_score(rfc_3, X_train_rfe, y_train, cv = 5, scoring= "balanced_accuracy"))

print(f"RFE: {rfe_score*100.0}%/{num_feats_rfe} feats. vs Baseline: {baseline*100.0}%/{num_feats_baseline} feats.")
```

```python
features_rfe = rfe.get_feature_names_out()
features_rfe
```

*Prácticamente el mismo resultado que con ANOVA (92,4%), algo peor que con todas las features y que con la selección visual sin colinealidades. Fíjate en que RFE y ANOVA eligen casi las mismas features, y varias son medidas casi equivalentes (radio, perímetro...): ninguno de los dos métodos tiene en cuenta la colinealidad entre features.*

*Otro aspecto importante es que RFE no usa ninguna métrica para seleccionar las features: en cada paso elimina la de menor importancia (`feature_importances_`) según el Random Forest.*

### Ejercicio 11

Prueba a eliminar features con el modelo de RandomForestClassifier (emplea SelectFromModel de sklearn). Utiliza esta vez una regresión logística para entrenar con las features escogidas y compara con el baseline. Muestra primero la importancia obtenida con el modelo baseline (que básicamente es el mismo que vamos a usar con SelectFromModel). Prueba primero para features por encima de la mediana y luego para las 5 primeras.

```python
#mostramos la importancia de las características del modelo original (sin reducción de características)
rfc.fit(X_train,y_train)
importances = rfc.feature_importances_
indices = np.argsort(importances)

plt.figure(figsize=(6, 6))
plt.title('Feature Importances')
plt.barh(range(len(indices)), importances[indices], color='b', align='center')
plt.yticks(range(len(indices)), [X_train.columns[i] for i in indices])
plt.xlabel('Relative Importance')
plt.show()
```

*Para las features superiores a la mediana*

```python
from sklearn.feature_selection import SelectFromModel
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

selector_model = SelectFromModel(estimator = rfc, threshold = "median")
selector_model.fit(X_train, y_train)
print("Escogidas:", selector_model.get_feature_names_out())
```

```python
from sklearn.linear_model import LogisticRegression

lc_clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter = 1000)) # La regresión logística necesita las features escaladas
X_train_model_median = selector_model.transform(X_train)
score_model_median = np.mean(cross_val_score(lc_clf, X_train_model_median, y_train, cv = 5, scoring = "balanced_accuracy"))
num_features_model_median = X_train_model_median.shape[1]
print(f"LR: {score_model_median*100.0}%/{num_features_model_median} feats. vs Baseline: {baseline*100.0}%/{num_feats_baseline} feats.")
```

*Para las 5 mejores según el Random Forest:*

```python
selector_model = SelectFromModel(estimator = rfc, threshold= "median", max_features= 5) 
selector_model.fit(X_train,y_train)
print("Escogidas:", selector_model.get_feature_names_out())

lc_clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter = 1000))
X_train_model_median = selector_model.transform(X_train)
score_model_median = np.mean(cross_val_score(lc_clf, X_train_model_median, y_train, cv = 5, scoring = "balanced_accuracy"))
num_features_model_median = X_train_model_median.shape[1]
print(f"LR: {score_model_median*100.0}%/{num_features_model_median} feats. vs Baseline: {baseline*100.0}%/{num_feats_baseline} feats.")
```

*Con las 15 features por encima de la mediana, la regresión logística (escalada) llega al 96,3%, por encima del baseline; con solo 5 baja al 93,1%. Ojo: aquí comparamos modelos distintos (regresión logística frente a Random Forest), así que la diferencia no se debe solo a la selección de features.*

### Ejercicio 12

Para terminar, evalúa contra test el modelo baseline (todas las features) y el de la selección que más te haya convencido. ¿Merece la pena la selección? Ten en cuenta tanto la métrica como el número de features.

```python
from sklearn.metrics import balanced_accuracy_score

X_test = test_set.drop(target, axis = 1)
y_test = test_set[target]

# Baseline: todas las features
rfc.fit(X_train, y_train)
score_test_baseline = balanced_accuracy_score(y_test, rfc.predict(X_test))

# Selección visual (sin colinealidades)
rfc_1.fit(X_train[features_visual_final], y_train)
score_test_visual = balanced_accuracy_score(y_test, rfc_1.predict(X_test[features_visual_final]))

print(f"Baseline: {score_test_baseline:.3f} con {X_train.shape[1]} features")
print(f"Selección visual: {score_test_visual:.3f} con {len(features_visual_final)} features")
```

*En test, la selección visual (8 features) consigue prácticamente lo mismo que el baseline con 30 (0,962 frente a 0,959). Así que sí merece la pena: un modelo con casi 4 veces menos features, más fácil de explicar y de mantener, sin perder rendimiento. Fíjate en el mensaje de fondo: seleccionar no ha mejorado claramente la métrica, pero ha simplificado mucho el modelo a coste prácticamente cero.*

### Ejercicio EXTRA:

Investiga RFECV y obtén el número óptimo de features siguiendo esta técnica, pero con la "balanced_accuracy" como métrica.

```python
# RFE con validación cruzada
from sklearn.feature_selection import RFECV

rfc_4 = RandomForestClassifier(random_state = 42)
rfecv = RFECV(estimator = rfc_4, step = 1, cv = 5, scoring = "balanced_accuracy", n_jobs = -1)
rfecv = rfecv.fit(X_train, y_train)

print("Número óptimo de features:", rfecv.n_features_)
print("Features seleccionadas:", list(rfecv.get_feature_names_out()))
print("Balanced accuracy en CV con esas features:", np.mean(cross_val_score(rfc_4, X_train[rfecv.get_feature_names_out()], y_train, cv = 5, scoring = "balanced_accuracy")))
```

*RFECV se queda con 23 features y una balanced accuracy en validación cruzada del 95,6%, por encima del baseline pero sin mejorar a la selección visual, que lo conseguía con solo 8. Además, en la gráfica siguiente verás que la curva es bastante plana a partir de unas 8 features (entre 0,94 y 0,96): muchas configuraciones son prácticamente equivalentes, así que el "óptimo" de RFECV puede cambiar bastante de una ejecución a otra.*

*Podemos observar la balanced accuracy en validación cruzada según el número de features:*

```python
# Balanced accuracy en validación cruzada vs número de features
plt.figure()
plt.xlabel("Número de features seleccionadas")
plt.ylabel("Balanced accuracy (CV)")
plt.plot(range(1, len(rfecv.cv_results_['mean_test_score']) + 1), rfecv.cv_results_['mean_test_score'])
plt.show()
```
